# Mrk 996

```python
import numpy as np
import astropy.units as u
```

```python
low_wav_obs = 4749.619 * u.angstrom
up_wav_obs = 9349.619 * u.angstrom

redshift = 0.00544 # James+2009
redshift = 0.0054 # Telles+2014

low_wav_rest = low_wav_obs / (1 + redshift)
up_wav_rest = up_wav_obs / (1 + redshift)

print(f"Observed wavelength range: {low_wav_obs:.2f} - {up_wav_obs:.2f}")
print(f"Rest wavelength range: {low_wav_rest:.2f} - {up_wav_rest:.2f}")
```
